# Oddball ERP Replication

**Purpose:** Reproduce the oddball ERP analysis across the walking conditions.

**Input:** Preprocessed EEG recordings and event annotations.

**Output:** Epochs from −0.2 to 0.8 s (baseline −0.2–0 s), evoked responses for standard and deviant stimuli, deviant–standard difference waves, and peak amplitude measurements (250–500 ms) at Cz/Pz.

```python
from pathlib import Path
import sys

def find_project_root(start: Path) -> Path:
    current = start.resolve()
    for candidate in [current] + list(current.parents):
        if (candidate / 'configs').exists() and (candidate / 'shared').exists():
            return candidate
    raise RuntimeError('Could not locate project root from notebook location.')

PROJECT_ROOT = find_project_root(Path.cwd())
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from shared.src.erp_replication import OddballERPConfig, run_oddball_replication
```

```python
CONFIG_PATH = PROJECT_ROOT / 'configs' / 'oddball_replication_authors.yaml'
config = OddballERPConfig.from_yaml(CONFIG_PATH, project_root=PROJECT_ROOT)
config
```

```python
outputs = run_oddball_replication(config)
outputs['run_manifest'].head()
```
