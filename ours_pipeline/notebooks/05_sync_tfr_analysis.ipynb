# Walking Synchronization Time–Frequency Analysis (Our Pipeline)

**Purpose:** Perform time–frequency analysis of EEG during the walking synchronization task using our preprocessing pipeline.

**Input:** Preprocessed EEG recordings with gait event markers.

**Output:** Stride-aligned epochs defined between consecutive heel strikes (RHS→RHS). Morlet wavelet transforms (3–35 Hz, 4–10 cycles) are computed and power is extracted for alpha-mu (7.5–12.5 Hz at C4/C6/CP4/CP6) and beta (16–32 Hz at C1/Cz/C2). Baseline correction is applied using −0.2 to 0 s before stride onset.
