# Shared preprocessing

This notebook runs the shared preprocessing engine after the data audit and event mapping steps.
It supports two profiles:
- **ours**: 0.1–40 Hz, notch 50/100 Hz, average reference, Picard-style ICA fallback.
- **authors**: 1–120 Hz, resample 250 Hz, average reference, infomax ICA, with ASR noted as optional.


```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve().parents[1]
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from shared.src.preprocessing import PreprocessingConfig, run_preprocessing
```

```python
CONFIG_PATH = PROJECT_ROOT / 'configs' / 'preprocessing_authors.yaml'
SUBJECT = None     # e.g. '001'
SESSION = None     # e.g. '01'
TASK = None        # e.g. 'act'
RUN = None         # e.g. '1'
MAX_RUNS = 1       # set to None to process all selected runs

config = PreprocessingConfig.from_yaml(CONFIG_PATH, project_root=PROJECT_ROOT)
config
```

```python
manifest = run_preprocessing(
    config,
    subject=SUBJECT,
    session=SESSION,
    task=TASK,
    run=RUN,
    max_runs=MAX_RUNS,
)
manifest.head()
```
