# Oddball ERP Analysis

**Purpose:** Compute ERP responses for the auditory oddball task across the walking conditions.

**Input:** Preprocessed EEG recordings and event annotations.

**Output:** Epochs extracted from −0.2 to 0.8 s with baseline correction (−0.2 to 0 s). Evoked responses for standard and deviant stimuli are computed, deviant–standard difference waves are generated, and peak amplitudes are measured for comparison across conditions.

```python
from pathlib import Path
import sys

def find_project_root(start: Path) -> Path:
    current = start.resolve()
    for candidate in [current] + list(current.parents):
        if (candidate / 'configs').exists() and (candidate / 'shared').exists():
            return candidate
    raise RuntimeError('Could not locate project root from notebook location.')

PROJECT_ROOT = find_project_root(Path.cwd())
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from shared.src.erp_replication import OddballERPConfig, run_oddball_replication
```

```python
CONFIG_PATH = PROJECT_ROOT / 'configs' / 'oddball_replication_ours.yaml'
config = OddballERPConfig.from_yaml(CONFIG_PATH, project_root=PROJECT_ROOT)
config
```
