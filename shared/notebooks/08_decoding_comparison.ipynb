**Output:** Summary comparisons of classification performance across subjects and analysis windows, illustrating how preprocessing differences influence decoding accuracy.

```python
from pathlib import Path
import sys

def find_project_root(start: Path) -> Path:
    current = start.resolve()
    for candidate in [current] + list(current.parents):
        if (candidate / 'configs').exists() and (candidate / 'shared').exists():
            return candidate
    raise RuntimeError('Could not locate project root from notebook location.')

PROJECT_ROOT = find_project_root(Path.cwd())
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
from shared.src.decoding_comparison import DecodingComparisonConfig, run_decoding_comparison
```

```python
CONFIG_PATH = PROJECT_ROOT / 'configs' / 'decoding_comparison.yaml'
config = DecodingComparisonConfig.from_yaml(CONFIG_PATH, project_root=PROJECT_ROOT)
outputs = run_decoding_comparison(config)
outputs['pipeline_summary']
```
